```python
# ============================================================
# TIC-TAC-TOE GAME
# ============================================================

# Function to display the board
def display_board(board):
    print()
    for i in range(3):
        print(" " + " | ".join(board[i]))
        if i < 2:
            print("---+---+---")
    print()


# Function to check whether a player has won
def check_win(board, player):

    # Check rows
    for i in range(3):
        if board[i][0] == board[i][1] == board[i][2] == player:
            return True

    # Check columns
    for i in range(3):
        if board[0][i] == board[1][i] == board[2][i] == player:
            return True

    # Check diagonals
    if board[0][0] == board[1][1] == board[2][2] == player:
        return True

    if board[0][2] == board[1][1] == board[2][0] == player:
        return True

    return False


# Function to check whether the board is full
def board_full(board):
    for row in board:
        if " " in row:
            return False
    return True


# Scores are maintained until players stop playing
x_score = 0
o_score = 0


while True:

    # Empty board represents the root/initial state
    board = [
        [" ", " ", " "],
        [" ", " ", " "],
        [" ", " ", " "]
    ]

    current_player = "X"

    print("\n========== TIC-TAC-TOE ==========")
    print("Enter row and column values from 1 to 3.")

    while True:

        display_board(board)

        # Find all possible moves from the current state
        possible_moves = []

        for i in range(3):
            for j in range(3):
                if board[i][j] == " ":
                    possible_moves.append((i, j))

        print("Possible Positions:",
              [(r + 1, c + 1) for r, c in possible_moves])

        print("\nPlayer", current_player, "Turn")

        # Get player's move
        try:
            row = int(input("Enter row (1-3): ")) - 1
            col = int(input("Enter column (1-3): ")) - 1

            # Check whether selected position is valid
            if (row, col) not in possible_moves:
                print("Invalid Move! Try again.")
                continue

        except ValueError:
            print("Please enter numbers only.")
            continue

        # Create the next game state
        board[row][col] = current_player

        print("\nPlayer", current_player,
              "selected position:", (row + 1, col + 1))

        # Check terminal state - Win
        if check_win(board, current_player):

            display_board(board)

            print("Player", current_player, "Wins!")
            print("+10 Points to Player", current_player)

            if current_player == "X":
                x_score += 10
            else:
                o_score += 10

            break

        # Check terminal state - Draw
        if board_full(board):

            display_board(board)

            print("Game is a Draw!")
            break

        # Switch player for the next state
        current_player = "O" if current_player == "X" else "X"


    # Display score after each game
    print("\n========== SCORE ==========")
    print("Player X:", x_score, "points")
    print("Player O:", o_score, "points")

    # Ask whether players want another game
    again = input("\nDo you want to play again? (yes/no): ").lower()

    if again != "yes":
        break


# Display final score
print("\n========== FINAL SCORE ==========")
print("Player X:", x_score, "points")
print("Player O:", o_score, "points")
print("Game Ended!")
```

```python
# ============================================================
# VACUUM CLEANER AGENT
# ============================================================

# Get the initial location of the vacuum cleaner
location = input("Enter Vacuum Location (A/B): ").upper()

# Get the initial condition of both rooms
room_A = input("Enter Status of Room A (Dirty/Clean): ").lower()
room_B = input("Enter Status of Room B (Dirty/Clean): ").lower()


# Display initial state
print("\n========== INITIAL STATE ==========")
print("Vacuum Location :", location)
print("Room A          :", room_A.capitalize())
print("Room B          :", room_B.capitalize())


step = 1

# Continue until both rooms become clean
while room_A == "dirty" or room_B == "dirty":

    print("\n---------- STEP", step, "----------")

    # Vacuum is currently in Room A
    if location == "A":

        print("Vacuum Location : Room A")

        # Clean Room A if it is dirty
        if room_A == "dirty":

            print("Room A Status   : Dirty")
            print("Action          : SUCK")

            room_A = "clean"

            print("Result          : Room A is now Clean")

        # If Room A is clean, move to Room B
        else:

            print("Room A Status   : Clean")
            print("Action          : Move RIGHT to Room B")

            location = "B"


    # Vacuum is currently in Room B
    elif location == "B":

        print("Vacuum Location : Room B")

        # Clean Room B if it is dirty
        if room_B == "dirty":

            print("Room B Status   : Dirty")
            print("Action          : SUCK")

            room_B = "clean"

            print("Result          : Room B is now Clean")

        # If Room B is clean, move to Room A
        else:

            print("Room B Status   : Clean")
            print("Action          : Move LEFT to Room A")

            location = "A"

    step += 1


# Display final goal state
print("\n========== FINAL STATE ==========")
print("Room A :", room_A.capitalize())
print("Room B :", room_B.capitalize())

print("\nGoal State Reached!")
print("Both rooms are clean.")
```
