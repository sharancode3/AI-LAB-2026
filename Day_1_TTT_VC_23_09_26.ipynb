```python
# TIC-TAC-TOE GAME

# Function to display the board
def display_board(board):
    print()
    for i in range(3):
        print(" " + " | ".join(board[i]))
        if i < 2:
            print("---+---+---")
    print()


# Function to check if a player has won
def check_win(board, player):

    # Check rows
    for i in range(3):
        if board[i][0] == board[i][1] == board[i][2] == player:
            return True

    # Check columns
    for i in range(3):
        if board[0][i] == board[1][i] == board[2][i] == player:
            return True

    # Check diagonals
    if board[0][0] == board[1][1] == board[2][2] == player:
        return True

    if board[0][2] == board[1][1] == board[2][0] == player:
        return True

    return False


# Function to check if the board is full
def board_full(board):
    for row in board:
        if " " in row:
            return False
    return True


# Initial scores
x_score = 0
o_score = 0


# Continue playing until the players choose to stop
while True:

    # Empty board represents the root state
    board = [
        [" ", " ", " "],
        [" ", " ", " "],
        [" ", " ", " "]
    ]

    current_player = "X"

    print("\n===== TIC-TAC-TOE =====")
    print("Enter row and column values from 1 to 3.")

    # Play one complete game
    while True:

        display_board(board)

        # Generate possible moves by finding empty positions
        possible_moves = []

        for i in range(3):
            for j in range(3):
                if board[i][j] == " ":
                    possible_moves.append((i, j))

        print("Possible positions:",
              [(r + 1, c + 1) for r, c in possible_moves])

        # Get current player's move
        print("\nPlayer", current_player)

        try:
            row = int(input("Enter row (1-3): ")) - 1
            col = int(input("Enter column (1-3): ")) - 1

            # Check whether the selected position is valid
            if (row, col) not in possible_moves:
                print("Invalid move! Try again.")
                continue

        except ValueError:
            print("Please enter numbers only.")
            continue

        # Create the next game state by making the move
        board[row][col] = current_player

        # Check terminal state - win
        if check_win(board, current_player):

            display_board(board)

            print("Player", current_player, "wins!")
            print("+10 points to Player", current_player)

            if current_player == "X":
                x_score += 10
            else:
                o_score += 10

            break

        # Check terminal state - draw
        if board_full(board):

            display_board(board)
            print("Game is a draw!")
            break

        # Move to the next level by switching the player
        if current_player == "X":
            current_player = "O"
        else:
            current_player = "X"

    # Display scores after every game
    print("\n===== SCORE =====")
    print("Player X:", x_score)
    print("Player O:", o_score)

    # Ask whether players want to continue
    again = input("\nDo you want to play again? (yes/no): ").lower()

    if again != "yes":
        break


# Final score when players stop playing
print("\n===== FINAL SCORE =====")
print("Player X:", x_score)
print("Player O:", o_score)

print("\nThanks for playing!")
```

```python
# Vacuum Cleaner Agent

# Get initial location of vacuum cleaner
location = input("Enter vacuum location (A/B): ").upper()

# Get status of both rooms
room_A = input("Enter status of Room A (Dirty/Clean): ").lower()
room_B = input("Enter status of Room B (Dirty/Clean): ").lower()

print("\n--- Initial State ---")
print("Vacuum Location:", location)
print("Room A:", room_A.capitalize())
print("Room B:", room_B.capitalize())

# Continue until both rooms are clean
while room_A == "dirty" or room_B == "dirty":

    # Vacuum is currently in Room A
    if location == "A":

        if room_A == "dirty":
            print("\nVacuum is in Room A")
            print("Room A is Dirty")
            print("Action: SUCK")

            room_A = "clean"
            print("Room A is now Clean")

        else:
            print("\nRoom A is Clean")
            print("Action: Move RIGHT to Room B")
            location = "B"

    # Vacuum is currently in Room B
    elif location == "B":

        if room_B == "dirty":
            print("\nVacuum is in Room B")
            print("Room B is Dirty")
            print("Action: SUCK")

            room_B = "clean"
            print("Room B is now Clean")

        else:
            print("\nRoom B is Clean")
            print("Action: Move LEFT to Room A")
            location = "A"


# Goal state
print("\n--- Final State ---")
print("Room A:", room_A.capitalize())
print("Room B:", room_B.capitalize())
print("Goal State Reached!")
print("Both rooms are clean.")
```
