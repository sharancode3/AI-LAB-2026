```python
# 8-Puzzle using Depth First Search (DFS)

def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


def get_successors(state):
    successors = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    for dr, dc, move in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col

            new_state = state.copy()
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            successors.append((new_state, move))

    return successors


def dfs(initial, goal):
    stack = [(initial, [], [initial])]
    visited = set()

    while stack:
        state, moves, path = stack.pop()

        state_tuple = tuple(state)

        if state_tuple in visited:
            continue

        visited.add(state_tuple)

        if state == goal:
            return moves, path

        successors = get_successors(state)

        # Reverse so RIGHT is explored first for this example
        for next_state, move in successors:
            if tuple(next_state) not in visited:
                stack.append(
                    (next_state,
                     moves + [move],
                     path + [next_state])
                )

    return None, None


# Initial State
initial = [
    1, 2, 3,
    0, 4, 6,
    7, 5, 8
]

# Goal State
goal = [
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
]


moves, path = dfs(initial, goal)

print("8-PUZZLE USING DFS\n")

print("Initial State:")
print_state(initial)

print("Goal State:")
print_state(goal)

if moves is not None:

    print("Solution Found!\n")

    for i in range(len(path)):
        print("State", i + 1)
        print_state(path[i])

    print("Solution Path:")
    print(" -> ".join(moves))

    print("\nNumber of Moves:", len(moves))

else:
    print("No Solution Found.")
```

```python
# 8-Puzzle using Iterative Deepening Search (IDS)

def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


def get_successors(state):
    successors = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    for dr, dc, move in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col

            new_state = state.copy()
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            successors.append((new_state, move))

    return successors


def depth_limited_search(state, goal, limit, path, moves, visited):

    if state == goal:
        return path, moves

    if limit == 0:
        return None

    for next_state, move in get_successors(state):

        state_tuple = tuple(next_state)

        if state_tuple not in visited:

            visited.add(state_tuple)

            result = depth_limited_search(
                next_state,
                goal,
                limit - 1,
                path + [next_state],
                moves + [move],
                visited
            )

            if result is not None:
                return result

            visited.remove(state_tuple)

    return None


def ids(initial, goal):

    depth = 0

    while True:

        print("Searching at Depth Limit:", depth)

        visited = {tuple(initial)}

        result = depth_limited_search(
            initial,
            goal,
            depth,
            [initial],
            [],
            visited
        )

        if result is not None:
            return result, depth

        depth += 1


initial = [
    1, 2, 3,
    0, 4, 6,
    7, 5, 8
]

goal = [
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
]


(result, depth) = ids(initial, goal)

path, moves = result

print("\nGoal State Found at Depth:", depth)

print("\nSolution Path:")

for i in range(len(path)):
    print("State", i + 1)
    print_state(path[i])

print("Moves:")
print(" -> ".join(moves))

print("\nNumber of Moves:", len(moves))
```
