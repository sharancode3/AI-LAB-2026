```python
import heapq

# Initial and Goal State
initial = (2, 8, 3,
           1, 6, 4,
           7, 0, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


# Heuristic: Number of Misplaced Tiles
def heuristic(state):
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1

    return count


# Generate possible moves
def get_neighbors(state):

    neighbors = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append(tuple(new_state))

    return neighbors


# Display puzzle
def display(state):

    for i in range(0, 9, 3):
        print(*state[i:i + 3])

    print()


# A* Algorithm
def a_star():

    # Priority queue: (f, g, state)
    open_list = []

    g = 0
    h = heuristic(initial)
    f = g + h

    heapq.heappush(open_list, (f, g, initial))

    parent = {initial: None}
    cost = {initial: 0}

    closed = set()

    while open_list:

        f, g, current = heapq.heappop(open_list)

        if current in closed:
            continue

        closed.add(current)

        # Goal test
        if current == goal:

            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()

            print("\nSolution Path:\n")

            for step, state in enumerate(path):

                h = heuristic(state)
                f = step + h

                print("Step:", step)
                print("g =", step, "h =", h, "f =", f)
                display(state)

            print("Goal State Reached!")
            print("Total Cost =", len(path) - 1)

            return

        # Generate successors
        for next_state in get_neighbors(current):

            if next_state in closed:
                continue

            new_g = g + 1

            if next_state not in cost or new_g < cost[next_state]:

                cost[next_state] = new_g
                parent[next_state] = current

                h = heuristic(next_state)
                f = new_g + h

                heapq.heappush(
                    open_list,
                    (f, new_g, next_state)
                )

    print("No solution found.")


# Main
print("A* Search - Misplaced Tiles")

print("\nInitial State:")
display(initial)

print("Goal State:")
display(goal)

a_star()
```

```python
import heapq

# Initial and Goal State
initial = (2, 8, 3,
           1, 6, 4,
           7, 0, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


# Heuristic: Manhattan Distance
def heuristic(state):

    distance = 0

    for i in range(9):

        tile = state[i]

        # Ignore blank
        if tile == 0:
            continue

        current_row = i // 3
        current_col = i % 3

        goal_index = goal.index(tile)

        goal_row = goal_index // 3
        goal_col = goal_index % 3

        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


# Generate possible moves
def get_neighbors(state):

    neighbors = []

    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append(tuple(new_state))

    return neighbors


# Display puzzle
def display(state):

    for i in range(0, 9, 3):
        print(*state[i:i + 3])

    print()


# A* Algorithm
def a_star():

    # Priority queue: (f, g, state)
    open_list = []

    g = 0
    h = heuristic(initial)
    f = g + h

    heapq.heappush(open_list, (f, g, initial))

    parent = {initial: None}
    cost = {initial: 0}

    closed = set()

    while open_list:

        f, g, current = heapq.heappop(open_list)

        if current in closed:
            continue

        closed.add(current)

        # Goal test
        if current == goal:

            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()

            print("\nSolution Path:\n")

            for step, state in enumerate(path):

                h = heuristic(state)
                f = step + h

                print("Step:", step)
                print("g =", step, "h =", h, "f =", f)
                display(state)

            print("Goal State Reached!")
            print("Total Cost =", len(path) - 1)

            return

        # Generate successors
        for next_state in get_neighbors(current):

            if next_state in closed:
                continue

            new_g = g + 1

            if next_state not in cost or new_g < cost[next_state]:

                cost[next_state] = new_g
                parent[next_state] = current

                h = heuristic(next_state)
                f = new_g + h

                heapq.heappush(
                    open_list,
                    (f, new_g, next_state)
                )

    print("No solution found.")


# Main
print("A* Search - Manhattan Distance")

print("\nInitial State:")
display(initial)

print("Goal State:")
display(goal)

a_star()
```
